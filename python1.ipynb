### Conditional statements(if/elif/else)

```python
score = 85

if score >= 90:
    print("Grade: A")
elif score >= 80:
    print("Grade: B")
elif score >= 70:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
# if statement : it ckeck whether the condition is true or false. If the condition is true, it executes the block of code inside the if statement. If the condition is false, it skips the block of code and moves on to the next statement.
# for example, in the above code, the condition is score >= 90. If the score is greater than or equal to 90, it will print "Grade: A". If the score is less than 90, it will check the next condition score >= 80. If the score is greater than or equal to 80, it will print "Grade: B". If the score is less than 80, it will check the next condition score >= 70. If the score is greater than or equal to 70, it will print "Grade: C". If the score is less than 70, it will print "Grade: F".
x = 9
if x == 9:
    print("x is equal to 9")
```

```python
# if-elif statement : it checks multiple conditions. If the first condition is true, it executes the block of code inside the if statement. If the first condition is false, it checks the next condition in the elif statement. If the second condition is true, it executes the block of code inside the elif statement. If the second condition is false, it checks the next condition in the elif statement. This process continues until a true condition is found or all conditions are checked. If no true condition is found, it executes the block of code inside the else statement.
# for e.g:
x = 32
if x > 50:
    print("x is greater than 50")
elif x > 30:
    print("x is greater than 30")
elif x > 20:
    print("x is greater than 20")
else:
    print("x is less than or equal to 20")
```

## Loops in Python
#### for loop : it is used to iterate over a sequence (list, tuple, string) or other iterable objects. It executes a block of code for each item in the sequence.


```python
# for e.g:
for i in range(5):
    print(i)
```

#### while loop : it is used to execute a block of code repeatedly as long as the condition is true. The condition is checked before the execution of the block of code. If the condition is true, the block of code is executed. If the condition is false, the loop is terminated and the program continues with the next statement after the loop.

```python
# for e.g:
i = 0
while i < 5:
    print(i)
    i += 1
```

## Break,pass and continue

### Break statement : it is used to terminate the loop when a certain condition is met. When the break statement is executed, the loop is immediately terminated and the program continues with the next statement after the loop.

```python
# for e.g:
numbers = [1, 2, 3, 4, 5]

for num in numbers:
    if num == 3:
        break  # Exit the loop entirely when num is 3
    print(num)

print("Loop ended.")

# Output:
# 1
# 2
# Loop ended.
```

#### pass statement : it is used as a placeholder for future code. When the pass statement is executed, nothing happens, but you avoid getting an error when empty code is not allowed.

```python
numbers = [1, 2, 3, 4, 5]

for num in numbers:
    if num == 3:
        pass  # Placeholder: do nothing, proceed as normal
    print(num)

# Output:
# 1
# 2
# 3
# 4
# 5
```

#### Continue statement : it is used to skip the current iteration of a loop and move on to the next iteration. When the continue statement is encountered, the rest of the code inside the loop for that particular iteration is skipped, and the loop proceeds to the next iteration.

```python
# for e.g:
for i in range(5):
    if i == 2:
        continue  # Skip the rest of the loop for this iteration
    print(i)
```

```python
a = float(input("Enter the first value: "))
b = float(input("Enter the second value: "))
operator = input("Enter the operator (+, -, *, /): ")

if operator == "+":
    total = a + b
    print(f"The sum of {a} and {b} is: {total}")
elif operator == "-":
    diff = a - b
    print(f"The difference between {a} and {b} is: { diff}")
elif operator == "*":
    print(f"The product of {a} and {b} is: {a * b}")
elif operator == "/":
    if b != 0:
        print(f"The division of {a} by {b} is: {a / b}")
    else:
        print("Error: Division by zero is not allowed.")  

```
