### Indexing and Slicing

```python
# In Python lists, indexing lets you access a single element by its position, while slicing lets you extract a continuous 
# portion (a sublist) of elements.
# Python uses zero-based indexing, meaning the first element is at index 0. Negative indices count backward from the end (-1 is the last item).
# Index (Positive):    0     1     2     3     4
# Elements:          ['a',  'b',  'c',  'd',  'e']
# Index (Negative):   -5    -4    -3    -2    -1
```

```python
fruits = ["apple", "banana", "cherry"]

fruits[0]        # 'apple'  (first item)
fruits[-1]       # 'cherry' (last item)
fruits[1:3]      # ['banana', 'cherry']  (slice: start inclusive, end exclusive)
fruits[:2]       # ['apple', 'banana']   (from start to index 2)
fruits[::-1]     # ['cherry', 'banana', 'apple']  (reversed)
```

### Common Methods

```python
fruits.append("mango")        # add to end
fruits.insert(1, "kiwi")      # insert at index
fruits.remove("banana")       # remove by value (first match)
fruits.pop()                  # remove & return last item
fruits.pop(0)                 # remove & return item at index 0
fruits.sort()                 # sort in place (ascending)
fruits.sort(reverse=True)     # sort descending
fruits.reverse()              # reverse in place
len(fruits)                   # length
"apple" in fruits             # True/False — membership check
fruits.count("apple")         # count occurrences
fruits.index("cherry")        # find index of value
```

# Tuples (when/why to use over lists).

```python
# TUPLES: A tuple is like a list but immutable — once created, it can't be changed.
# Use tuples when data shouldn't change (coordinates, RGB values, fixed records).
# ordered sequence of elements.
# Tuples are constructed using parentheses ()
```

### Key properties of Tuples

```python
# Ordered	 :    Elements retain their exact insertion order and are zero-indexed.
# Immutable  :  	Once created, elements cannot be added, removed, or modified.
# Heterogeneous :	Can contain mixed data types (integers, strings, lists, functions, etc.).
# Allows Duplicates :	Duplicate values are permitted since items are accessed by position.
# Hashable  : 	If all elements inside a tuple are immutable, the tuple itself is hashable (can be used as a dictionary key or set element).
```

```python
# Empty tuple
empty = ()
print(empty)
# Standard tuple
point = (10, 20)
print(point)
# Without parentheses (Tuple Packing)
person = "Alice", 30, "Engineer"
print(person)
# Tuple constructor
converted = tuple([1, 2, 3])  # Converts list to tuple (1, 2, 3)
print(converted)
```

```python
# Why use a tuple over a list?
# 1.Slightly faster and uses less memory
# 2.Signals intent — "this data is fixed"
# 3.Can be used as dictionary keys (lists can't, because they're mutable)
```

```python
# Single-Element Tuple (Trailing Comma Requirement)
# To create a tuple with only one element, you must include a trailing comma. 
# Without it, Python evaluates the expression as a standard parenthesized value.
not_a_tuple = ("hello")   # Evaluates to <class 'str'>
is_a_tuple  = ("hello",)  # Evaluates to <class 'tuple'>
print(type(not_a_tuple))
print(type(is_a_tuple))
```

### Accessing Elements (Indexing & Slicing)

```python
# Tuples support the exact same zero-based indexing and slicing mechanics as lists.
data = ("a", "b", "c", "d", "e")

# Positive and Negative Indexing
print(data[0])   # 'a'
print(data[-1])  # 'e'

# Slicing [start:stop:step]
print(data[1:4])  # ('b', 'c', 'd')
print(data[::-1]) # ('e', 'd', 'c', 'b', 'a') (Reversed)
```

### Immutability & The "Mutable Element" Caveat Attempting to Modify a Tuple

```python
# Modifying, adding, or deleting an item directly in a tuple raises a TypeError.
coordinates = (12.5, 45.2)
print(coordinates)
# coordinates[0] = 15.0  --> TypeError: 'tuple' object does not support item assignment
```

#### Tuples Containing Mutable Objects

```python
# While the tuple structure itself cannot change, mutable objects inside a tuple can still be modified.
# Tuple containing a list
nested = (1, 2, [10, 20])

# Modifying the internal list is allowed
nested[2].append(30)
print(nested)  # Output: (1, 2, [10, 20, 30])
```

#### Tuple Operations & Unpacking

```python
# Tuple Unpacking
# Unpacking assigns elements of a tuple to individual variables in a single line:
# Basic Unpacking
point = (3, 4)
x, y = point  # x = 3, y = 4

# Extended Unpacking using * (Asterisk operator)
numbers = (1, 2, 3, 4, 5)
first, *middle, last = numbers

print(first)   # 1
print(middle)  # [2, 3, 4] (returns a list)
print(last)    # 5
```

```python
# Tuple Concatenation & Repetition
# Combining tuples produces a new tuple object in memory:
t1 = (1, 2)
t2 = (3, 4)

print(t1 + t2)  # (1, 2, 3, 4)
print(t1 * 3)   # (1, 2, 1, 2, 1, 2)
```

```python
# Built-in Tuple Methods
# Tuples only have two built-in methods because they are immutable:
values = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)

# 1. count(x) -> Returns the number of times x occurs
print(values.count(7))  # Output: 2

# 2. index(x) -> Returns the first zero-based index of x
print(values.index(8))  # Output: 3
```

# Tuples Vs Lists 

```python
# Mutability: Lists are mutable (can be modified, appended, or resized in place). Tuples are immutable (cannot be altered after creation).

# Syntax: Lists use square brackets [1, 2, 3]. Tuples use parentheses (1, 2, 3).

# Memory Efficiency: Tuples consume less memory because they have a fixed size. 
# Lists allocate extra memory overhead to handle dynamic resizing.

# Performance Speed: Tuples are faster to create and iterate through than lists due to their fixed size and C-level memory optimization.

# Dictionary Keys & Sets: Tuples are hashable (provided their elements are immutable) and can be used as dictionary keys or set elements.
#  Lists are unhashable and cannot.

# Built-in Methods: Lists have many modification methods (append(), insert(), pop(), remove(), sort()). 
# Tuples only have two methods (count() and index()).

# Semantic Use Case: Lists represent homogeneous, dynamic sequences of items that change over time. 
# Tuples represent heterogeneous, fixed structure records (e.g., coordinate pairs (x, y) or database rows).
```

# Dictionaries (keys, values, methods, iteration)

```python
# Dictionaries :A dictionary in Python is an unordered (or insertion-ordered since Python 3.7+), mutable collection of key-value pairs.
# Dictionaries are optimized for retrieving values when key names are known.
# Syntax: Enclosed in curly braces {key: value}.
# The Key is like the word you look up.
# The Value is like the definition attached to that word.
# Lookup Speed: Accessing items by key takes $O(1)$ constant time on average due to underlying hash tables.
```

### The Key (The Identifier)

```python
# 1.Uniqueness: Every key in a dictionary must be unique. If you assign a value to an existing key, it overwrites the old value.
# 2.Immutability (Hashability): Keys must be of an immutable data type so their internal hash value never changes.
#        >> Allowed as keys: str, int, float, bool, tuple (if it contains only immutable items).
#        >> NOT allowed as keys: list, dict,
```

### The Value (The Content)

```python
# 1.No Type Restrictions:Values can be any Python data type—integers, strings, floats, lists, functions, custom objects,or even another dictionary.
# 2.Duplicates Allowed: Multiple keys can hold the exact same value.
# 3.Mutability: Values can be modified, replaced, or updated freely.
```

```python
# Standard curly braces
user = {"name": "Alice", "age": 30, "role": "Engineer"}

# Using dict() constructor with keyword arguments
user = dict(name="Alice", age=30, role="Engineer")

# Using dict() with a list of key-value tuples
user = dict([("name", "Alice"), ("age", 30)])
```

## Accessing & Modifying Values

```python
# Accessing Items
# Using square brackets raises a KeyError if the key does not exist. The .get() method returns None
# (or a specified default) if the key is missing.
person = {"name": "Bob", "score": 95}

# Bracket notation
print(person["name"])  # 'Bob'
# print(person["age"])

# .get() method (safe because it wouldnot crash if key is missing.)
print(person.get("score"))       # 95
print(person.get("rank"))        # None
print(person.get("rank", 0))     # 0 (custom default)
```

```python
# Modifying and Adding Items
# Dictionaries are mutable, allowing values to be updated or new pairs inserted in place.
person["score"] = 98     # Updates existing key
person["city"] = "NYC"   # Adds new key-value pair
print(person)
```

```python
# Removing Elements >>>

# 1.pop(key) :	Removes specified key and returns.
data = {"a": 1, "b": 2, "c": 3}
data.pop("b")
print(data)

# 2.popitem() :	Removes and returns the last inserted (key, value) tuple.
data = {"a": 1, "b": 2, "c": 3}
data.popitem()
print(data)
# del dict[key] :	Deletes key directly (raises KeyError if missing).
data = {"a": 1, "b": 2, "c": 3}
del data["b"]
print(data)

# clear() :	Removes all items from the dictionary.
data = {"a": 1, "b": 2, "c": 3}
data.clear()
print(data)
```

```python
user = {"name": "Alice", "role": "admin", "age": 30}

# Removes 'role' and stores its value in a variable
role = user.pop("role")

print(role)  # Output: 'admin'
print(user)  # Output: {'name': 'Alice', 'age': 30}
```

### Dictionary Iteration or looping 

```python
data = {"a": 1, "b": 2, "c": 3}

# Iterating over Keys (default behavior)
for key in data:
    print(key)

# Iterating over Values
for val in data.values():
    print(val)

# Iterating over Key-Value Pairs
for key, val in data.items():
    print(f"{key} -> {val}")
```

#### Dictionary Comprehension

```python

# Similar to list comprehensions, dictionary comprehensions build new dictionaries cleanly:
# Square numbers from 1 to 5
squares = {x: x**2 for x in range(1, 6)}
# Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Filter even values from an existing dict
prices = {"apple": 1.5, "banana": 0.8, "cherry": 2.5}
cheap = {item: price for item, price in prices.items() if price < 2.0}
# Output: {'apple': 1.5, 'banana': 0.8}
```

#### Merging Dictionaries

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 99, "c": 3}  # 'b' overlaps

# Merge operator (|) -> creates a new dict
merged = dict1 | dict2
print(merged)  # {'a': 1, 'b': 99, 'c': 3} (dict2 overrides dict1)

# Update in place using .update() or |=
dict1.update(dict2)
```

# Sets

```python
# Sets : A set in Python is an unordered, mutable collection of unique, hashable (immutable) elements.
# Sets are modeled after mathematical sets and are optimized for membership testing and eliminating duplicates.
nums = {1, 2, 3, 3, 2}   # becomes {1, 2, 3}(duplicates are not allowed.)
# Standard set creation
fruits = {"apple", "banana", "cherry"}

# Automatic deduplication
numbers = set([1, 2, 2, 3, 4, 4, 4])
print(numbers)  # Output: {1, 2, 3, 4}

# IMPORTANT: Empty set creation
empty_set = set()      # Correct <class 'set'>
empty_dict = {}        # Incorrect! Creates an empty dict <class 'dict'>
```

```python
# properties of sets:
# Syntax: Enclosed in curly braces {1, 2, 3} or created using set().
# Uniqueness: Automatically removes duplicate values.
# Unordered: Items have no fixed sequence or position (no indexing or slicing like s[0]).
# Lookup Speed: Checking if an item is in a set (x in s) takes $O(1)$ constant time on average using hash tables.
# Mathematical Operations: Supports union ($\cup$), intersection ($\cap$), difference ($-$), and symmetric difference ($\Delta$).
```

### Unique & Unordered

```python
s = {3, 1, 2, 1, 3}
print(s)  # Output: {1, 2, 3} (Duplicates removed; order is arbitrary)
```

### Elements Must Be Hashable (Immutable)

```python
# Set elements themselves must be immutable (strings, numbers, tuples). You cannot put mutable objects like lists or dictionaries inside a set.
valid_set = {1, "hello", (1, 2)}      # Allowed
# invalid_set = {1, [1, 2]}           # Raises TypeError: unhashable type: 'list'
```

### Adding and Removing Elements

```python
# add(x) :	Adds single item x to the set.(s.add(5))
set1 = {1,2,5,6,5,8}
set1.add(9)
print(set1)

# update(iterable) :	Adds all multiple items from an iterable. (s.update([10, 20]))
set1 = {1,2,5,6,5,8}
set1.update([20,4,54,3])
print(set1)

# discard(x) :	Removes x; does not raise an error if missing.(safe remove)(s.discard(99))
set1 = {1,2,5,6,5,8}
(set1.discard(8))
print(set1)

# remove(x) :	Removes x; raises KeyError if missing.(Strict Remove)(s.remove(5))
set1 = {1,2,5,6,5,8}
set1.remove(2)
print(set1)

# pop()	: Removes and returns an arbitrary element.(item = s.pop())
set1 = {1,2,5,6,5,8}
item = set1.pop()
print(set1)

# clear() :	Empties the set completely.(s.clear())
set1 = {1,2,5,6,5,8}
set1.clear()
print(set1)
```

#### Difference (--)

```python
print(A - B)               # {1, 2}
print(A.difference(B))     # Equivalent
```

```python
print(A ^ B)                         # {1, 2, 5, 6}
print(A.symmetric_difference(B))     # Equivalent
```

###  Contact book (dictionary-based)

```python
# Build a CLI-style program (can just be a loop with input()) that supports:

# Add — store as {"name": {"phone": ..., "email": ...}} (dict of dicts)
# Search — look up by name, use .get() so it doesn't crash on missing names
# Delete — remove an entry, handle the case where it doesn't exist
# # Bonus: List all contacts sorted alphabetically
```

```python
def contact_book():
    contacts = {}

    while True:
        print("\n" + "=" * 30)
        print("    CONTACT BOOK MANAGER")
        print("=" * 30)
        print("1. Add Contact")
        print("2. Search Contact")
        print("3. Delete Contact")
        print("4. List All Contacts (Sorted)")
        print("5. Exit")

        choice = input("Enter your choice(1-5):").strip()

        if choice == "1":
            name = input("Enter contact name:")
            if not name :
                print("name required ")
                continue
            phone = (input(("Enter contact phone number :")))
            email = input("Enter contact email:")
            contacts[name]={
            "phone" : phone,
            "email" : email
            }
            print(f"contact {name} updated successfully!")
        elif choice == "2":
            name = input("Enter name to search: ")

            info = contacts.get(name)

            if info:
                print(f"Detail for {name}:")
                print(f"phone : {info['phone']}")
                print(f"email : {info['email']}")
            else:
                print(f"contact {name} not found!")

        elif choice == "3":
            name = input("Enter name to delete:")

            Removed_contact =contacts.pop(name)
            if Removed_contact:
                print(f"Contact {name} deleted successfully!")
            else:
                print(f"Cannot delete,Contact {name} not found!")

        elif choice == "4":
            if not contacts:
                print("contact book empty!")
            else:
                print("Contact Lists(Alphabatically):")

                for name in sorted(contacts.keys()):
                    details = contacts[name]
                    print(f"     - {name}")
                    print(f"     - Phone: {details['phone']}")
                    print(f"     - Email: {details['email']}")

        elif choice == "5":
            print("Exiting Contact Book Manager. Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number from 1 to 5.")

if __name__ == "__main__":
   contact_book()
```

### Word frequency counter

```python
# Take a string (a sentence or paragraph)
# Split it into words (.split())
# Loop through and build a dictionary: {word: count}
# Hint: word_counts[word] = word_counts.get(word, 0) + 1 is the classic one-liner pattern here — try writing it the long way with if/else first so you understand it, then try the shortcut.
# Bonus: ignore case and punctuation (.lower(), and strip punctuation using string.punctuation)
```

```python
import string
text = input("Enter the string: ")
    
words = text.lower().split()

word_counts = {}

for word in words:
    cleaned_word = word.strip(string.punctuation)
    
    if not cleaned_word:
        continue  # Skip if string becomes empty after removing punctuation
        
    if cleaned_word in word_counts:
        word_counts[cleaned_word] += 1
    else:
        word_counts[cleaned_word] = 1

print(word_counts)
```
